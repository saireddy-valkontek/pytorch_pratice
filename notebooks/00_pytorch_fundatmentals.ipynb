### range and like tensors

```python
one_to_ten = torch.arange(start=1,end=100,step=10)
one_to_ten
```

```python
one_to_ten.shape, one_to_ten.ndim
```

```python
ten_zeros = torch.zeros_like(input=one_to_ten)
```

```python
ten_zeros
```

```python
ten_ones = torch.ones_like(input=one_to_ten)
```

```python
ten_ones
```
