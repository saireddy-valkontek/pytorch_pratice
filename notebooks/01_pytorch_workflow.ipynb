```python
weight , bias = 0.7,0.3
```

```python
train_split = int(0.8*len(X))
x_train,y_train = X[:train_split],y[:train_split]
x_test , y_test = X[train_split:],y[train_split:]
```

```python
def plot_predictions(train_data = x_train, train_labels=y_train,
                    test_data=x_test, test_labels=y_test,
                    predictions=None):
    plt.figure(figsize=(8,6))
    plt.scatter(train_data,train_labels, c="b", s=4, label= "Training Date") # c= color , s= size
    plt.scatter(test_data, test_labels, c="g", s=4, label="Test Data")
    
    if predictions is not None:
        plt.scatter(test_data, predictions, c="r",s=4, label="Predictions")

    plt.legend(prop={"size":14})
   
```

```python
plot_predictions()
```

#### pytroch model for linear regression

```python
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.weights = nn.Parameter(torch.randn(1,
                                                requires_grad=True,
                                                dtype=torch.float))
        self.bias = nn.Parameter(torch.randn(1,
                                            requires_grad=True,
                                            dtype=torch.float))
    # forward method
    def forward(self, x:torch.Tensor) -> torch.Tensor: 
        return self.weights*x + self.bias
```

```python
torch.manual_seed(42)
model_0 = LinearRegressionModel()
model_0 , model_0.parameters()
```

```python
epochs = 10 # an one epoch is one loop thourgh the data
```

```python
import matplotlib.pyplot as plt

weight_history = []
grad_history = []

for epoch in range(epochs):

    model_0.train()

    y_pred = model_0(x_train)

    loss_value = loss(y_pred, y_train)

    optimizer.zero_grad()

    loss_value.backward()

    # store values
    weight_history.append(model_0.weights.item())
    grad_history.append(model_0.weights.grad.item())

    optimizer.step()

# plot weights
plt.figure(figsize=(8, 4))
plt.plot(weight_history)
plt.xlabel("Epoch")
plt.ylabel("Weight Value")
plt.title("Weight Change Over Epochs")
plt.show()

# plot gradients
plt.figure(figsize=(8, 4))
plt.plot(grad_history)
plt.xlabel("Epoch")
plt.ylabel("Gradient Value")
plt.title("Gradient Change Over Epochs")
plt.show()
```
