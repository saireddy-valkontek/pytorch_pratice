```python
x_reshape = x.reshape(1,10)
x_reshape , x_reshape.shape, x_reshape.ndim
```

```python
x_reshape = x.reshape(9,1)
x_reshape , x_reshape.shape, x_reshape.ndim
```

the reshaped tensor when multiplied need to the actuatl shape of the original tenosr

```python
x = torch.arange(1.,10.)
x, x.shape, x.ndim
```

```python
z = x.view(1,9)
z, z.shape
```

view points to the orginal tensor so changing z changes the x they share the same memoery

```python
x_stacked = torch.stack([x,x,x,x])
x_stacked, x_stacked.shape, x_stacked.ndim
```

```python
x_stacked = torch.stack([x,x,x,x],dim=1)
x_stacked, x_stacked.shape, x_stacked.ndim
```
